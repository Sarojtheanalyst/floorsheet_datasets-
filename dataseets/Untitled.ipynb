```python
import os
all_datas=os.listdir()
```

```python
import pandas as pd
```

```python
all_merged=[]
for item in all_datas:
    if item.endswith(".csv"):
        df=pd.read_csv(item)
        all_merged.append(df)
print("loaded all the data sucessfully ")
```

```python
len(all_merged)
```

```python
all_merged[1]
```

```python
final_datasets=pd.concat(all_merged,ignore_index=False)
final_datasets
```

```python
final_datasets.to_csv("merged_datasets.csv",index=False)
```
